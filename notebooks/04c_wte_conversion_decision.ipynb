# **Phase 4C: Cluster-Informed WtE Conversion Decision**
---
**Research question:** Given a biomass sample belonging to fuel-cluster X,
with process-suitability constraints known, which conversion pathway /
technology is most appropriate?

**Decision layering**
- Clusters define *what the fuel is*
- Suitability rules define *what the fuel can do*
- Decision rules define *what we should actually build*


```python
# load data
import pandas as pd
df = pd.read_csv("../data/interim/engineered_features_with_cluster.csv")
```

```python
df.head()
```

```python
# apply phase 4A(process suitability)
import sys, os
sys.path.append(os.path.abspath(".."))

from src.models.suitability_rules import apply_process_rules

df = apply_process_rules(df)
```

```python
# apply phase 4C (conversion rules)
from src.models.wte_conversion_rules import assign_conversion_technology

df = assign_conversion_technology(df)

# save output
df.to_csv(
    "../data/processed/wte_final_decisions.csv",
    index=False
)
```

```python
# Result inspection
tech_dist = df['Final_Conversion_Technology'].value_counts()
tech_dist

# save output
tech_dist.to_csv(
    "../results/tables/phase4c/wte_technology_distribution.csv"
)
```

```python
summary = pd.crosstab(
    df["Cluster"],
    df["Final_Conversion_Technology"],
    normalize="index"
)
summary

# save output
summary.to_csv(
    "../results/tables/phase4c/cluster_vs_wte_technology.csv"
)
```

**Decision layer summary**

| Layer | Description |
|---|---|
| Cluster | Data-driven biomass grouping |
| Primary process | Physics-based feasibility |
| Constraint level | Operational realism |
| Final conversion | Engineering decision |
